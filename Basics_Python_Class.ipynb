```python
#Predictive analytics 
#1 Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
#2. load the data
data = pd.read_csv("student_data.csv")
data.tail()
```

```python
data.isnull
```

```python
prediction = []

for index, student in data.iterrows():
    average_score = (student["assignment_pct"] + student["quiz_pct"])/2
    if student["attendance_pct"] < 70 or average_score < 60:
        prediction.append("Unlikely to pass")
    else:
        prediction.append("Likely to pass")
#prediction
```

```python
data["average_score"] = (data["assignment_pct"] + data["quiz_pct"])/2
data["prediction"] = prediction
data.tail
```

```python
#3. Assign input and output variables
X = data[["quiz_pct"]] #Inport Variable
y = data["average_score"] #Output Variable
```

```python
import matplotlib.pyplot as plt
#plt.figure(figure(10,6))
plt.scatter(X, y)
plt.xlabel("Quiz (%)", fontsize = 14, fontweight = "bold") 
plt.ylabel("Average Scores (%)", fontsize = 14, fontweight = "bold")
plt.grid(True, axis = "y")
plt.tight_layout()
plt.show()
```

```python
#4 Define a Model 
model = LinearRegression()
model_fit = model.fit(X, y)
```

```python
predicted_data = model_fit.predict(X)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, y, label = "Raw Data")
plt.plot(X, predicted_data, color = "red", linewidth = 3, label = "Regression Line")
plt.xlabel("Quiz (%)", fontsize = 14, fontweight= "bold")
plt.ylabel("Average Scores (%)", fontsize = 14, fontweight = "bold")
plt.grid(True, axis = "y")
plt.title("Predictive Analytics of Average Scores Dependant on Quiz Score")
plt.legend()
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression

df = pd.DataFrame()

df["hours_studied"] = [1, 2, 3, 4, 5]
df["exam_scores"] = [50, 55, 65, 70, 80]

df.to_csv("study_scores.csv", index=False)

# Step 3: Read the CSV file
data = pd.read_csv("study_scores.csv")

# Step 4: Define input and output variables
X = data[["hours_studied"]]
y = data["exam_scores"]

# Step 5: Create and train the regression model
model = LinearRegression()
model.fit(X, y)

# Step 6: Predict exam scores
predicted_data = model.predict(X)

print("Predicted hour of study:", predicted_data)
```
